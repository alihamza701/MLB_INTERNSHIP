```python
from google.colab.patches import cv2_imshow
import cv2

image1 = cv2.imread("/content/image2.jpeg")

cv2_imshow(image1)

print("\n\n**Dimensions**")
print("Hieght : " , image1.shape[0])
print("Width : " , image1.shape[1])
print("\nChannels : " , image1.shape[2])
print("\nFile size : " , image1.size)
```

```python
import os

grayscale = cv2.cvtColor(image1,cv2.COLOR_BGR2GRAY)

cv2_imshow(grayscale)

output_dir = "/content/output_files/"

os.makedirs(output_dir, exist_ok=True)

cv2.imwrite(output_dir + "grayscale.png" , grayscale)
```

```python
resized  = cv2.resize(image1 , (287,200))

cv2_imshow(resized)

cv2.imwrite(output_dir + "resized.png" , resized)
```

```python
crop1 = image1[0:100,0:100]
crop2 = image1[100:300,100:300]
crop3 = image1[100:287 , 400:600]

cv2_imshow(crop1)
print("\n\n")
cv2_imshow(crop2)
print("\n\n")
cv2_imshow(crop3)

cv2.imwrite(output_dir + "crop1.png" , crop1)
cv2.imwrite(output_dir + "crop2.png" , crop2)
cv2.imwrite(output_dir + "crop3.png" , crop3)
```

```python
rotate90 = cv2.rotate(resized , cv2.ROTATE_90_CLOCKWISE)
rotate180 = cv2.rotate(resized , cv2.ROTATE_180)
rotate270 = cv2.rotate(resized , cv2.ROTATE_90_COUNTERCLOCKWISE)

cv2_imshow(rotate90)
print("\n\n")
cv2_imshow(rotate180)
print("\n\n")
cv2_imshow(rotate270)

cv2.imwrite(output_dir + "rotate90.png" , rotate90)
cv2.imwrite(output_dir + "rotate180.png" , rotate180)
cv2.imwrite(output_dir + "rotate270.png" , rotate270)
```

```python
horizontally = cv2.flip(resized , 1)
vertically = cv2.flip(resized , 0)

cv2_imshow(horizontally)
print("\n\n")
cv2_imshow(vertically)
cv2.imwrite(output_dir + "horizontally_flipped.png" , horizontally)
cv2.imwrite(output_dir + "vertically_flipped.png"  , vertically)
```

```python
img_for_shapes = cv2.imread("/content/image1.jpeg")

rectangle  = cv2.rectangle(img_for_shapes , (400,400),(200,200), (0,200,0) , 5)
cv2_imshow(rectangle)
```

```python
cv2.circle(img_for_shapes , (400,200) , 40 , (100,200,255) , 5)

cv2_imshow(img_for_shapes)
```

```python
cv2.line(img_for_shapes , (200,600) ,(300, 100) , (255,255,255) , 4 )
cv2_imshow(img_for_shapes)

#Hieght :  487
#Width :  740
```

```python
import cv2
import numpy as np

points = np.array([
    [100, 100],
    [200, 50],
    [300, 100],
    [270, 200],
    [130, 200]
], np.int32)

points = points.reshape((-1, 1, 2))

cv2.polylines(
    img_for_shapes,
    [points],
    True,
    (255, 0, 255),
    4
)

cv2_imshow(img_for_shapes)

cv2.imwrite(output_dir + "shapesDrawn.png" , img_for_shapes)
```
