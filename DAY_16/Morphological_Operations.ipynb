```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("/content/image2.jpeg")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

kernel = np.ones((5, 5), np.uint8)

blurred = cv2.GaussianBlur(gray, (5, 5), 0)
canny = cv2.Canny(blurred, 50, 150)

erosion = cv2.erode(canny, kernel, iterations=1)
dilation = cv2.dilate(canny, kernel, iterations=1)
opening = cv2.morphologyEx(canny, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(canny, cv2.MORPH_CLOSE, kernel)
gradient = cv2.morphologyEx(canny, cv2.MORPH_GRADIENT, kernel)
tophat = cv2.morphologyEx(canny, cv2.MORPH_TOPHAT, kernel)
blackhat = cv2.morphologyEx(canny, cv2.MORPH_BLACKHAT, kernel)

plt.figure(figsize=(16, 8))
plt.suptitle("Morphological Operations - Before vs After", fontsize=14)

plt.subplot(2, 4, 1)
plt.imshow(canny, cmap="gray")
plt.title("Before (Canny)")
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(erosion, cmap="gray")
plt.title("Erosion")
plt.axis("off")

plt.subplot(2, 4, 3)
plt.imshow(dilation, cmap="gray")
plt.title("Dilation")
plt.axis("off")

plt.subplot(2, 4, 4)
plt.imshow(opening, cmap="gray")
plt.title("Opening")
plt.axis("off")

plt.subplot(2, 4, 5)
plt.imshow(canny, cmap="gray")
plt.title("Before (Canny)")
plt.axis("off")

plt.subplot(2, 4, 6)
plt.imshow(closing, cmap="gray")
plt.title("Closing")
plt.axis("off")

plt.subplot(2, 4, 7)
plt.imshow(gradient, cmap="gray")
plt.title("Gradient")
plt.axis("off")

plt.subplot(2, 4, 8)
plt.imshow(tophat, cmap="gray")
plt.title("Top Hat")
plt.axis("off")

plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 4))
plt.imshow(blackhat, cmap="gray")
plt.title("Black Hat")
plt.axis("off")
plt.tight_layout()
plt.show()
```
